```python
import os
import pandas as pd 
import numpy as np
file_dir = r'YOURFILEDIR\data'
file_path = os.path.join(file_dir, "export.csv")
df = pd.read_csv(file_path)
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# 1. Convert dates to datetime
df['Date Rptd'] = pd.to_datetime(df['Date Rptd'])
df['DATE OCC'] = pd.to_datetime(df['DATE OCC'])

# 2. Fill and convert numeric code columns from float to int
code_columns = ['Crm Cd 2', 'Crm Cd 3', 'Crm Cd 4', 'Weapon Used Cd']

for col in code_columns:
    df[col] = df[col].fillna(0).astype('int64')
```

```python
df['TIME OCC'] = (
    df['TIME OCC']
    .astype(str)
    .str.replace(r'\.0$', '', regex=True)
    .str.zfill(4) 
)
df['TIME OCC'] = df['TIME OCC'].str.slice(0,2) + ':' + df['TIME OCC'].str.slice(2,4)
```

```python
df['DATE OCC'] = pd.to_datetime(df['DATE OCC']).dt.date
df['Date Rptd'] = pd.to_datetime(df['Date Rptd']).dt.date
```

```python
df[['DATE OCC', 'Date Rptd']].head(10)
```

```python
df['Mocodes'] = df['Mocodes'].fillna('')
```

```python
 df['Vict Sex'].unique()
```

```python
df['Vict Sex'] = df['Vict Sex'].replace([r'^\s*$', '-'], np.nan, regex=True)
sex_mapping = {
    'M' : 'Male',
    'F' : 'Female',
    'X' : 'Unknown',
    'H' : 'Unknown'
}
df['Vict Sex'] = df['Vict Sex'].map(sex_mapping).fillna('Not Applicable / Unknown')
```

```python
 df['Vict Descent'].unique()
```

```python
df['Vict Descent'] = df['Vict Descent'].replace([r'^\s*$', '-'], np.nan, regex=True)
descent_mapping = {
    'A': 'Other Asian',
    'B': 'Black',
    'C': 'Chinese',
    'D': ' Cambodian',
    'F': 'Filipino',
    'G': 'Guamanian',
    'H': 'Hispanic/Latin/Mexican',
    'I': 'American Indian/Alaskan Native',
    'J': 'Japanese',
    'K': 'Korean',
    'L': 'Laotian',
    'O': 'Other',
    'P': 'Pacific Islander',
    'S': 'Samoan',
    'U': 'Hawaiian',
    'V': 'Vietnamese',
    'W': 'White',
    'X': 'Unknown',
    'Z': 'Asian Indian'
}
df['Vict Descent'] = df['Vict Descent'].map(descent_mapping).fillna('Unknown / Not Specified')

```

```python
df['Vict Descent'].value_counts(dropna=False)
```

```python
valid_premises = df.dropna(subset=['Premis Cd', 'Premis Desc'])
premis_dict = dict(zip(valid_premises['Premis Cd'], valid_premises['Premis Desc']))


df['Premis Desc'] = df['Premis Desc'].fillna(df['Premis Cd'].map(premis_dict))

df['Premis Desc'] = df['Premis Desc'].fillna('UNKNOWN')
df['Premis Cd'] = df['Premis Cd'].fillna('UNKNOWN')
```

```python
df.isnull().sum()
```

```python
df['Weapon_Used_Flag'] = (df['Weapon Used Cd'] != 0).astype(int)
```

```python
df['Weapon Used Cd'] = df['Weapon Used Cd'].fillna(0)
```

```python
df['Weapon Desc'] = df['Weapon Desc'].fillna('NO WEAPON / NOT APPLICABLE')
```

```python
 print(df[df['Status'].isnull()][['Status', 'Status Desc']])
```

```python
df['Status Desc'].unique()
```

```python
#Creating a lookup
valid_status = df.dropna(subset=['Status', 'Status Desc'])
status_reverse_dict = dict(zip(valid_status['Status Desc'], valid_status['Status']))

df['Status'] = df['Status'].fillna(df['Status Desc'].map(status_reverse_dict))
```

```python
filtered_df = df[df['Crm Cd 1'].isnull()][[ 'Crm Cd' ,'Crm Cd Desc', 'Crm Cd 1',]]
print(filtered_df)
```

```python
valid_status = df.dropna(subset=['Crm Cd', 'Crm Cd 1'])
status_reverse_dict = dict(zip(valid_status['Crm Cd 1'], valid_status['Crm Cd']))

df['Crm Cd 1'] = df['Crm Cd 1'].fillna(df['Crm Cd']).map(status_reverse_dict)
```

```python
df[['Crm Cd 2', 'Crm Cd 3', 'Crm Cd 4',]] = df[['Crm Cd 2', 'Crm Cd 3', 'Crm Cd 4',]].fillna(0) 
```

```python
df['Cross Street'] = df['Cross Street'].fillna('None')
```

```python
df['LOCATION'] = df['LOCATION'].str.replace(r'\s+', ' ', regex=True).str.strip()
```

```python
df['DR_NO'].duplicated().sum()
```

```python
# Replace multiple consecutive spaces with a single space and strip edges for CROSS STREET
df['Cross Street'] = df['Cross Street'].str.replace(r'\s+', ' ', regex=True).str.strip()
```

```python
df = df.rename(columns={
    'DR_NO' : 'record_no',
    'DATE OCC' : 'date_occurred',
    'Date Rptd': 'date_reported',
    'TIME OCC': 'time_occurred',
    'AREA' : 'area',
    'AREA NAME' : 'area_name',
    'Rpt Dist No' : 'reporting_district_no',
    'Part 1-2' : 'part_1-2',
    'Crm Cd' : 'crime_code',
    'Crm Cd Desc' : 'crime_code_desc',
    'Mocodes' : 'mocodes',
    'Vict Age' : 'vict_age',
    'Vict Sex' : 'vict_sex',
    'Vict Descent' : 'vict_descent',
    'Premis Cd' : 'premis_code',
    'Premis Desc' : 'premis_desc',
    'Weapon Used Cd' : 'weapon_used_code',
    'Weapon Desc' : 'weapon_desc',
    'Status' : 'status',
    'Status Desc' : 'status_desc',
    'Crm Cd 1' : 'crime_code_1',
    'Crm Cd 2' : 'crime_code_2',
    'Crm Cd 3' : 'crime_code_3',
    'Crm Cd 4' : 'crime_code_4',
    'LOCATION' : 'location',
    'Cross Street': 'cross_street',
    'LAT' : 'lat',
    'LON' : 'lon',
    'Weapon_Used_Flag' : 'weapon_used_flag'
})
```

```python
# Exporting To SSMS 
import pandas as pd
from sqlalchemy import create_engine

server = r"SERVERNAME\SQLEXPRESS"
database = "CrimeRecords"

connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&trusted_connection=yes"
    "&TrustServerCertificate=yes"
)

engine = create_engine(connection_string)
```

```python
with engine.connect() as conn:
    print("Connected successfully!")
```

```python
df.to_sql(
    "lapd_crime_records",
    con=engine,
    schema="dbo",
    if_exists="replace",
    index=False)
```
